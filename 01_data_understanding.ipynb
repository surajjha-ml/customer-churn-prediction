This notebook focuses on understanding the structure of the IBM Telco Customer

Special attention is given to temporal and target leakage to ensure that
model performance reflects real-world deployment behavior.


## Dataset Overview

This project uses the **IBM Telco Customer Churn** dataset.

Each row represents a customer account, with features describing:
- Customer demographics
- Services subscribed
- Account and contract details
- Billing information
- Churn outcome

The dataset is treated as a snapshot of customer state prior to a churn decision.

## Snapshot Assumption

The Telco dataset does not contain explicit timestamps.

Therefore, we assume:
- Each row represents a **snapshot of customer state**
- All feature values are observed **before** the churn decision
- The `Churn` column represents the outcome occurring after this snapshot

This assumption is common in industry when working with static customer tables.


## Target Variable

The target variable is `Churn`:

- Churn = "Yes" → Customer churned
- Churn = "No" → Customer retained

This is a binary classification problem focused on predicting churn risk.


## Target Leakage Check

The `Churn` column must be excluded from:
- Feature engineering
- Scaling
- Model training

Any feature derived directly or indirectly from `Churn` would result
in target leakage.


## Potential Leakage Risks in Telco Data

In churn datasets, leakage often arises from:
- Account status indicators
- Cancellation or termination flags
- Post-churn billing adjustments
- Features reflecting final customer actions

Although this dataset is relatively clean, each feature will be evaluated
for whether it would realistically be available at prediction time.


## High-Risk Columns Review

The following columns require careful interpretation:
- `TotalCharges`
- `tenure`
- Contract-related features

These features appear historical but must be validated to ensure they do not
encode post-churn information.


## Temporal Leakage Considerations

Because the dataset is provided as a static snapshot:
- Explicit temporal leakage is limited
- The primary risk is **implicit leakage** through feature semantics

This makes careful feature interpretation more important than timestamp filtering.


## Implications for Train / Validation Split

Due to the absence of timestamps:
- Random splitting will be used
- Leakage risk is mitigated through feature vetting rather than time-based splits

This limitation is acknowledged and documented.


## Assumptions and Limitations

Assumptions:
- Feature values represent customer state prior to churn
- No post-churn behavior is encoded in predictors

Limitations:
- No explicit time dimension
- Churn window cannot be directly controlled

These limitations are common in real-world legacy datasets.
