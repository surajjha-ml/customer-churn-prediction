# Customer Churn Prediction — Business Understanding

## Project Context
This project aims to build a **customer churn prediction model** to support real-world
retention decisions.

This is a **production-style ML project**, not an academic exercise.
The emphasis is on:
- Clear business framing
- Actionable predictions
- Cost-sensitive decision-making
- Prevention of data leakage

The model is designed to help prioritize **retention efforts** for active customers.


## Business Problem Statement

Customer acquisition is expensive, and retaining existing customers is often significantly
more cost-effective than acquiring new ones.

However, retention teams typically operate under:
- Limited budgets
- Limited operational capacity
- Incomplete visibility into customer intent

**Problem:**  
The business cannot proactively intervene for all customers and must identify
which customers are most likely to churn in the near future.

**Goal:**  
Predict the probability that an active customer will churn in the near future,
so that retention actions can be targeted where they are most likely to be effective.


## What Is Customer Churn?

In this project, *churn* refers to a customer who was previously active but stops
using the service or generating revenue within a defined future time window.

Churn is treated as a **forward-looking event**, meaning:
- Predictions are made for currently active customers
- Labels are assigned based on future behavior


## Possible Churn Definitions

In practice, churn definitions vary by industry and business model.

Common approaches include:

1. **Contractual churn**  
   - Customer cancels or does not renew a subscription

2. **Behavioral churn**  
   - Customer becomes inactive for a specified number of days

3. **Revenue churn**  
   - Customer stops generating revenue within a future window

Each definition involves trade-offs between clarity, timeliness, and business relevance.


## Chosen Churn Definition

For this project, churn is defined as:

> **An active customer who stops using the service (or paying) within the next 30 days.**

Rationale:
- Short enough to allow proactive intervention
- Commonly used in real-world retention teams
- Easy to explain to non-technical stakeholders
- Avoids overcomplication while remaining realistic


## Prediction Setup

Predictions are made using a **snapshot-based approach**:

- Each customer is observed at a specific snapshot date (time T)
- Only information available **at or before time T** is used
- Churn is determined by observing customer behavior in the **30-day period after T**

This setup ensures:
- Predictions are forward-looking
- The model reflects how it would behave in production
- Temporal data leakage can be explicitly controlled


## Target Variable

The target variable is binary:

- `churn = 1`  
  Customer churns within 30 days after the snapshot date

- `churn = 0`  
  Customer remains active beyond the 30-day window

This label definition aligns model outputs with real retention decisions.


## Customer Scope

### In Scope
- Customers who are active at the snapshot date
- Customers eligible for retention actions

### Out of Scope
- Customers who have already churned
- Customers with insufficient history (if applicable)
- One-time or transient users, depending on business context

Restricting the population ensures that predictions are actionable
and avoids label contamination.


## Business Costs and Trade-offs

Churn prediction is a **cost-sensitive** problem.

- **False Negative (FN):**
  - Customer is predicted to stay but actually churns
  - Results in lost revenue and lifetime value

- **False Positive (FP):**
  - Customer is predicted to churn but stays
  - Results in unnecessary retention cost (e.g., discounts, outreach)

In most churn scenarios:
- False negatives are more costly than false positives
- The optimal decision threshold depends on retention budget and capacity


## Model Success Metrics

### Primary Metric
- **ROC-AUC**
  - Measures the model’s ability to rank customers by churn risk
  - Independent of any single threshold

### Business Evaluation
- Threshold-based analysis to compare:
  - Aggressive vs conservative retention strategies
  - Trade-offs between retention cost and churn reduction

Accuracy alone is not a meaningful metric for this problem.


## Model Output and Usage

The model produces a **churn probability score** for each active customer.

This score can be used to:
- Rank customers by churn risk
- Define risk segments (e.g., high / medium / low)
- Allocate retention resources more effectively

The model supports decision-making rather than replacing human judgment.


## Out of Scope for This Project

This project intentionally does not:
- Make causal claims about why customers churn
- Optimize retention campaigns end-to-end
- Model full customer lifetime value (LTV)

The focus is on building a **reliable, explainable churn risk model**
that can realistically be deployed to support retention teams.
